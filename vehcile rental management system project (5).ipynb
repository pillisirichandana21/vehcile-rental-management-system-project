```python

                                            #VEHICLE RENTAL MANAGEMENT SYSTEM

from abc import ABC, abstractmethod
import csv


# ---------------- ABSTRACT CLASS ----------------

class Vehicle(ABC):

    def __init__(self, vehicle_id, name, price_per_day):

        # Protected attributes
        self._vehicle_id = vehicle_id
        self._name = name

        # Private attribute
        self.__price_per_day = price_per_day

        self._available = True

    # Getter
    def get_price(self):
        return self.__price_per_day

    # Setter
    def set_price(self, price):
        self.__price_per_day = price

    def get_id(self):
        return self._vehicle_id

    def is_available(self):
        return self._available

    def rent(self):
        self._available = False

    def return_vehicle(self):
        self._available = True

    @abstractmethod
    def display_details(self):
        pass


# ---------------- CAR CLASS ----------------

class Car(Vehicle):

    def __init__(self, vehicle_id, name, price_per_day):
        super().__init__(vehicle_id, name, price_per_day)

    def display_details(self):

        status = "Available" if self._available else "Rented"

        print(
            self._vehicle_id,
            "|",
            self._name,
            "| Rs.",
            self.get_price(),
            "per day |",
            status
        )


# ---------------- RENTAL SYSTEM CLASS ----------------

class RentalSystem:

    def __init__(self):
        self.vehicles = []
        self.rentals = []

    # Add vehicle
    def add_vehicle(self):

        try:

            vehicle_id = input("Enter Vehicle ID: ")
            name = input("Enter Vehicle Name: ")

            price = float(
                input("Enter Rental Price Per Day: ")
            )

            if price <= 0:
                raise ValueError(
                    "Price must be greater than zero."
                )

            vehicle = Car(
                vehicle_id,
                name,
                price
            )

            self.vehicles.append(vehicle)

            print("Vehicle added successfully.")

        except ValueError as e:
            print("Invalid input:", e)

    # Display vehicles
    def display_vehicles(self):

        if len(self.vehicles) == 0:
            print("No vehicles available.")
            return

        print("\n----- VEHICLES -----")

        for vehicle in self.vehicles:
            vehicle.display_details()

    # Rent vehicle
    def rent_vehicle(self):

        vehicle_id = input(
            "Enter Vehicle ID to rent: "
        )

        for vehicle in self.vehicles:

            if vehicle.get_id() == vehicle_id:

                if not vehicle.is_available():
                    print("Vehicle is already rented.")
                    return

                try:

                    customer = input(
                        "Enter Customer Name: "
                    )

                    days = int(
                        input("Enter Number of Days: ")
                    )

                    if days <= 0:
                        raise ValueError(
                            "Days must be greater than zero."
                        )

                    charge = (
                        vehicle.get_price() * days
                    )

                    vehicle.rent()

                    self.rentals.append(
                        [
                            vehicle_id,
                            customer,
                            days,
                            charge
                        ]
                    )

                    print("\nVehicle rented successfully.")
                    print("Customer:", customer)
                    print("Number of Days:", days)
                    print("Total Charge: Rs.", charge)

                except ValueError as e:
                    print("Invalid input:", e)

                return

        print("Vehicle not found.")

    # Return vehicle
    def return_vehicle(self):

        vehicle_id = input(
            "Enter Vehicle ID to return: "
        )

        for vehicle in self.vehicles:

            if vehicle.get_id() == vehicle_id:

                if vehicle.is_available():
                    print(
                        "This vehicle is not currently rented."
                    )
                    return

                vehicle.return_vehicle()

                print(
                    "Vehicle returned successfully."
                )

                return

        print("Vehicle not found.")

    # Save rental details
    def save_rentals(self):

        try:

            with open(
                "rentals.csv",
                "w",
                newline=""
            ) as file:

                writer = csv.writer(file)

                writer.writerow(
                    [
                        "Vehicle ID",
                        "Customer",
                        "Days",
                        "Charge"
                    ]
                )

                writer.writerows(self.rentals)

            print(
                "Rental records saved successfully."
            )

        except IOError:
            print("Error while saving rental file.")


# ---------------- MAIN PROGRAM ----------------

system = RentalSystem()

while True:

    print("\n===== VEHICLE RENTAL MANAGEMENT =====")

    print("1. Add Vehicle")
    print("2. Display Vehicles")
    print("3. Rent Vehicle")
    print("4. Return Vehicle")
    print("5. Save Rentals")
    print("6. Exit")

    try:

        choice = int(
            input("Enter your choice: ")
        )

        if choice == 1:
            system.add_vehicle()

        elif choice == 2:
            system.display_vehicles()

        elif choice == 3:
            system.rent_vehicle()

        elif choice == 4:
            system.return_vehicle()

        elif choice == 5:
            system.save_rentals()

        elif choice == 6:

            system.save_rentals()

            print("Thank you!")
            break

        else:
            print(
                "Please enter choice between 1 and 6."
            )

    except ValueError:

        print(
            "Please enter a valid number."
        )
```
